```python
import os
import subprocess
import glob
import shutil

# Define output directories
output_dir = '/content/rnaseq_outputs'
fastqc_dir = os.path.join(output_dir, 'fastqc')
salmon_dir = os.path.join(output_dir, 'salmon_quants')
os.makedirs(fastqc_dir, exist_ok=True)
os.makedirs(salmon_dir, exist_ok=True)

print("Starting Quality Control and Read Mapping...")

# Loop through the metadata to process each sample
for index, row in metadata.iterrows():
    sample_name = row['SampleName']

    # Use recursive globbing to find the fastq files inside subfolders
    search_pattern = os.path.join(folder_path, "**", f"*{sample_name}*.f*q.gz")
    files = sorted(glob.glob(search_pattern, recursive=True))

    if not files:
        print(f"Could not find fastq files for {sample_name} in subfolders of {folder_path}")
        continue

    print(f"\n[{index+1}/{len(metadata)}] Processing {sample_name}...")

    # 1. Run FastQC
    print("  -> Running FastQC...")
    fastqc_cmd = ['fastqc', '-q', '-t', '2', '-o', fastqc_dir] + files
    subprocess.run(fastqc_cmd)

    # 2. Run Salmon Mapping
    print("  -> Running Salmon mapping...")
    salmon_out = os.path.join(salmon_dir, sample_name)
    if len(files) == 1:
        # Single-end mapping
        salmon_cmd = ['salmon', 'quant', '-i', 'celegans_transcripts_index', '-l', 'A',
                      '-r', files[0], '-p', '2', '--validateMappings', '-o', salmon_out]
    elif len(files) >= 2:
        # Paired-end mapping (assuming first two files are R1 and R2)
        salmon_cmd = ['salmon', 'quant', '-i', 'celegans_transcripts_index', '-l', 'A',
                      '-1', files[0], '-2', files[1], '-p', '2', '--validateMappings', '-o', salmon_out]

    # Run salmon silently (capture output) to avoid clogging the notebook
    subprocess.run(salmon_cmd, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

# 3. Aggregate QC results with MultiQC
print("\nRunning MultiQC to aggregate reports...")
subprocess.run(['multiqc', output_dir, '-o', output_dir, '-q'])

# 4. Zip the results for local download
print("\nZipping results for download...")
zip_path = '/content/rnaseq_results_archive'
shutil.make_archive(zip_path, 'zip', output_dir)

print(f"\nDone! All outputs are zipped successfully.")
print(f"You can now download '{zip_path}.zip' from the files panel on the left (folder icon).")
```
